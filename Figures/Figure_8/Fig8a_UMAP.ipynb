```python
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
import scipy.stats as stats
import scanpy as sc
import scanpy.external as sce
```

```python
sc.set_figure_params(dpi=300, dpi_save=300, color_map = 'viridis')
sc.settings.verbosity = 3
```

```python
cd4 = sc.read_h5ad("../../Pass2_Annotation/Lifespan_PCA_CD4T.h5ad")
```

```python
naivecd4 = cd4[cd4.obs['Annotation'].isin(['CD4_naive_SOX4+','CD4_naive_SOX4-'])].copy()
```

```python
sc.tl.umap(naivecd4)
```

```python

sc.pl.umap(naivecd4, color=['Annotation'], palette=['#a5ce39','#1d3a1f',], save='NaiveCD4T.pdf')
```

```python
sc.pl.umap(naivecd4, color=['SOX4'], cmap='Reds', vmax=2.5, save='NaiveCD4T_SOX4.pdf')
```
