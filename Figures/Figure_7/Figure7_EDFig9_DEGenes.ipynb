```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from matplotlib.pyplot import rc_context
```

```python
CD4_DE = pd.read_csv("./NaiveCD4_DE.csv", index_col=0)
CD8_DE = pd.read_csv("./NaiveCD8_DE.csv", index_col=0)
```

```python
CD4_logcpm = pd.read_csv("./NaiveCD4_CPM_corrected.csv", index_col=0)
CD8_logcpm = pd.read_csv("./NaiveCD8_CPM_corrected.csv", index_col=0)
```

# Figure 7g and h

```python
g = sb.clustermap(CD4_logcpm.loc[list(CD4_DE[CD4_DE['logFC'] > 0.3].index)+list(CD4_DE[CD4_DE['logFC'] < -0.3].index)], z_score=0, cmap='RdYlBu_r', col_cluster=False, row_cluster=False,figsize=(15,30), vmin=-3, vmax=3)
g.savefig("./figures/CD4_SOX4_DE_Heatmap.pdf")
```

```python
g = sb.clustermap(CD8_logcpm.loc[list(CD8_DE[CD8_DE['logFC'] > 0.3].index)+list(CD8_DE[CD8_DE['logFC'] < -0.3].index)], z_score=0, cmap='RdYlBu_r', col_cluster=False, row_cluster=False,figsize=(15,30), vmin=-3, vmax=3)
g.savefig("./figures/CD8_SOX4_DE_Heatmap.pdf")
```

# Extended Data Figure 9i and j

```python
numnaivecd4_up = len(CD4_DE[CD4_DE['logFC'] > 0.3])
numnaivecd4_down = len(CD4_DE[CD4_DE['logFC'] < -0.3])

fig = plt.figure(figsize=(3,3))
plt.title("Naive CD4+ T")
sb.barplot(x=["Upregulated SOX4", "Downregulated"],y=[numnaivecd4_up, numnaivecd4_down], palette=['#a3cd3a','#203a21'])
plt.ylabel('# of DE Genes')
plt.xticks(rotation=45)
plt.grid(False)
plt.ylim([0,50])

fig.savefig("./figures/NCD4_SOX4_Barplot.pdf")

print("Upregulated: "+str(numnaivecd4_up))
print("Downregulated: "+str(numnaivecd4_down))
```

```python
numnaivecd8_up = len(CD8_DE[CD8_DE['logFC'] > 0.3])
numnaivecd8_down = len(CD8_DE[CD8_DE['logFC'] < -0.3])

fig = plt.figure(figsize=(3,3))
plt.title("Naive CD8+ T")
sb.barplot(x=["Upregulated SOX4", "Downregulated"],y=[numnaivecd8_up, numnaivecd8_down], palette=['#ffd6a2','#f47525'])
plt.ylabel('# of DE Genes')
plt.xticks(rotation=45)
plt.grid(False)
plt.ylim([0,50])
fig.savefig("./figures/NCD4_SOX8_Barplot.pdf")


print("Upregulated: "+str(numnaivecd8_up))
print("Downregulated: "+str(numnaivecd8_down))
```
